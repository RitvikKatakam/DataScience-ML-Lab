OOPS

* Encapsulation
* abstraction
* inheritence
* polymorphism

```python
class BankAccount():
  def __init__(self, name, balance, age):  # self means the current object
    self.name = name                       # name is the property in the class
    self.__balance = balance               # balance is private (encapsulation)
    self.age = age                         # fixed assignment

  def deposit(self, amount):
    self.__balance += amount               # use __balance instead of balance

  def withdraw(self, amount):
    if amount <= self.__balance:
      self.__balance -= amount
    else:
      print("Insufficient balance")

  def get_balance(self):
    return self.__balance
```

```python
acc = BankAccount("Ritvik", 1000, 22)

acc.withdraw(100)

print("Current balance:", acc.get_balance())

acc.deposit(500)

print("Current balance after withdrawing:", acc.get_balance())
```
