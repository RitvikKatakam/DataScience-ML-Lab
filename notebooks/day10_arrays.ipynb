```python
a = [1,2,3,4,5,6]
print(a)
```

```python
# access elements
print("element in 0 index is",a[0])
print("element in 3rd index is",a[3])
```

```python
# modify elements
print("before modifying:",a)
a[5]=10
print("after modifying:",a)
```

```python
#append
a.append(20)
print(a)
```

```python
#insert
a.insert(2,30)
a
```

```python
# delete
a.remove(3)
a
```

```python
#traversing an array
for i in a:
  print(i)
```
