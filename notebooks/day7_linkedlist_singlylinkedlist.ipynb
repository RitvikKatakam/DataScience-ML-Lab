#linked list

* linked list is a linear data structure where we store reference to the object
* linked list consists of various nodes
* each node consists of two things  

```python
class Node():
  def __init__(self,data):
    self.data=data
    self.next=None

class Singlylinkedlist(): #nodes only a point to the next node,can't go back to the previos node
  def __init__(self):
    self.head=None

  def insertatbeginning(self,data):
    new_node=Node(data)
    new_node.next=self.head
    self.head=new_node

  def insertatend(self,data):
    new_node=Node(data)
    if self.head is None:
      self.head=new_node
      return
    temp = self.head # Corrected indentation
    while temp.next:
      temp=temp.next

    temp.next=new_node

  def display(self):
    current = self.head # Use a local variable to traverse
    while current:
      print(current.data,'next_node',current.next)
      current = current.next # Move to the next node

    print(None)
```

```python
obj1 = Singlylinkedlist()
obj1.insertatbeginning(10)
```

```python
obj1.display()
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class SinglyLinked2:
    def __init__(self):
        self.head = None

    # Insert at beginning
    def insert_first(self, data):
        new_node = Node(data)
        new_node.next = self.head   # connect to old head
        self.head = new_node        # move head

    # Insert at end
    def insert_end(self, data):
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            return

        temp = self.head
        while temp.next:
            temp = temp.next

        temp.next = new_node

    # Display
    def display2(self):
        temp = self.head
        while temp:
            if temp.next:
                print(f"{temp.data} -> {temp.next.data}")
            else:
                print(f"{temp.data} -> None")
            temp = temp.next
```

```python
obj = SinglyLinked2()

obj.insert_first(10)
obj.insert_first(20)
obj.insert_end(30)

obj.display2()
```
