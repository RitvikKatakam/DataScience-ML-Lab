#Control Flow statement
These statement are executed based on the certain conditions are met.
* If Statement:




```python
name="venkat"
if name == "venkat":
  print("my name is venkat")
```

# if-else


```python
name="ritvik"
if name == "venkat":
  print("my name is venkat")
else:
  print("my name is not venkat")
```

# else-if condition


```python
marks =59
if marks>=90:
  print("Grade A")
elif marks>=80:
  print("grade b")
elif marks>=70:
  print("Grade C")
else:
  print("fail")
```

# For loop
it used to iterate over an item or list
if item is null or empty for loop never be printed

```python
n=10
for i in range(0):
  print(i)
```

```python
for i in range(10):
  if i%2!=0:
    continue

  print(i)
```

continue:skip the current iteration and move to the other iteration

```python
for i in range(10):
  if i%2!=0:
    break

  print(i)
```

#pass
pass is a null statement and it does nothing

```python
for i in range(10):
  if i%2!=0:
    pass

  print(i)
```

#while
Run the loop till the condition is true

```python
names=[]
while len(names) < 4:
  value=input("enter a name:")
  names.append(value)
print(names)
```

```python
day = 5

match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case _:
        print("Invalid day")
```
