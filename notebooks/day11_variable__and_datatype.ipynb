```python
class test:
  def __init__(self,val1,val2):
    self.val1 = val1
    self.val2 = val2

  def display(self):
    print(self.val1)
    print(self.val2)

  def swap(self):
    self.val1,self.val2=self.val2,self.val1
    print(self.val1)
    print(self.val2)
```

```python
testobj = test(10,20)
print("before swapping")
testobj.display()
print("after swapping")
testobj.swap()
```
