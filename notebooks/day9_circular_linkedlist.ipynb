Circular linked list

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class CircularLinkedList:
    def __init__(self):
        self.head = None

    # here pass only insert end and display function
    def insert_node(self, data):
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            new_node.next = self.head
            return

        temp = self.head

        # move to last node
        while temp.next != self.head:
            temp = temp.next

        temp.next = new_node
        new_node.next = self.head

    def display(self):
        if self.head is None:
            print("List is empty")
            return

        temp = self.head
        while True:
            print(temp.data, end=" ")
            temp = temp.next
            if temp == self.head:
                break
        print("back to head")
```

```python
cll = CircularLinkedList()

cll.insert_node(10)
cll.insert_node(20)
cll.insert_node(30)

cll.display()
```

Doublely circular linked list

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.prev = None
        self.next = None


class DoublyCircularLinkedList:
    def __init__(self):
        self.head = None

    def insert_node(self, data):
        new_node = Node(data)

        # if list is empty
        if self.head is None:
            self.head = new_node
            new_node.next = new_node
            new_node.prev = new_node
            return

        last = self.head.prev

        last.next = new_node
        new_node.prev = last

        new_node.next = self.head
        self.head.prev = new_node

    def display(self):
        if self.head is None:
            print("List is empty")
            return

        temp = self.head
        while True:
            print(temp.data, end=" ")
            temp = temp.next
            if temp == self.head:
                break
        print("back to head")
```

```python
cdll=DoublyCircularLinkedList()

cdll.insert_node(10)
cdll.insert_node(20)
cdll.insert_node(30)

cdll.display()
```
