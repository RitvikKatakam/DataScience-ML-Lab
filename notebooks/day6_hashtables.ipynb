Hash is a data structure that stores key:value pair.

It provides very fast lookup, insertion, and deletion.

Python has already a built in hash like structure called Dictionary.

Hash function converts a key into index.

Storage: It is array based.

Keys are converted to indices.

```python
class Hash():
    def __init__(self, size=10):
        self.size = size
        self.table = [[] for _ in range(self.size)]

    def _hash(self, key):
        return hash(key) % self.size

    def insert(self, key, value):
        index = self._hash(key)

        for i in range(len(self.table[index])):
            if self.table[index][i][0] == key:
                self.table[index][i] = (key, value)
                return

        self.table[index].append((key, value))

    def get_value(self, key):
        index = self._hash(key)

        for item in self.table[index]:
            if item[0] == key:
                return item[1]

        return None

    def display(self):
        for index, item in enumerate(self.table):
            print(index, ":", item)
```

```python
ht = Hash()

ht.insert("name", "cam")
ht.insert("role", "AI")

print(ht.get_value("name"))
ht.display()
```
