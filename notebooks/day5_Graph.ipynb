a graph is a data structure to represent relantionship between objects

vertices-nodes(object)

edges-connect between the object

1.straight forward graph which is undirected

```python
from collections import defaultdict
```

```python
class undirected_graph:
  def __init__(self):
    self.graph=defaultdict(list)

  def add_edge(self,v1,v2):
    self.graph[v1].append(v2)
    self.graph[v2].append(v1)

  def display(self):
    for vertex,neighbours in self.graph.items():
      print(vertex,'->',neighbours)
```

```python
myobject = undirected_graph()
myobject.add_edge('hyderabad','banglore')
myobject.add_edge('banglore','chennai')
myobject.display()
```

2. directed graph

```python
class directed_graph:
  def __init__(self):
    self.graph=defaultdict(list)
  def add_edge(self,v1,v2):
    self.graph[v1].append(v2)
  def display(self):
    for vertex,neighbours in self.graph.items():
      print(vertex,'follows',neighbours)
```

```python
my_object = directed_graph()
my_object.add_edge('sam','vaibhav')
my_object.display()
```
