Undirected graph adajency list

```python
class undirectedgraph:
    def __init__(self, vertices):
        self.vertices = vertices
        self.graph = [[0 for column in range(vertices)] for row in range(vertices)]
    def addEdge(self,v1,v2):
      self.graph[v1][v2] = 1
      self.graph[v2][v1] = 1
    def display(self):
      for i in range(self.vertices):
        print(i,"->",self.graph[i])
```

```python
g = undirectedgraph(3)
g.addEdge(0,1)
g.addEdge(0,2)

g.display()
```

DFS for directed graph

First what is directed graph, u can call it a one way graph

dfs - depth for search

go deep as possible

then backtrack

best examples is course prequisites

```python
class DFS:
    def __init__(self):
        self.graph = {}
        self.visited = set()

    def addEdge(self, u, v):
        if u not in self.graph:
            self.graph[u] = []
        self.graph[u].append(v)

    def dfs(self, node):
        self.visited.add(node)
        print(node, end=" ")

        for neighbor in self.graph.get(node, []):
            if neighbor not in self.visited:
                self.dfs(neighbor)
```

```python
g = DFS()

g.addEdge("Math", "Programming")
g.addEdge("Programming", "Data Structures")
g.addEdge("Data Structures", "Algorithm")
g.addEdge("Algorithm", "AI")

print("DFS Traversal (Course Order):")
g.dfs("Math")
```

weighted unidirected graph

```python
class Wightedundirectedgraph:
  def __init__(self):
     self.graph = {}
  def addEdge(self,v1,v2,weight):
    if v1 not in self.graph:
      self.graph[v1] = {}
    if v2 not in self.graph:
      self.graph[v2] = {}
    self.graph[v1][v2] = weight
    self.graph[v2][v1] = weight

  def display(self):
    for i in self.graph:
      print(i,"->",self.graph[i])
```

```python
g=Wightedundirectedgraph()
g.addEdge(0,1,10)
g.addEdge(0,2,3)
g.display()
```
