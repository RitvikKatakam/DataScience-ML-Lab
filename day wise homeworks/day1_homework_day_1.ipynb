Student Grades

Create a list of 5 student scores.

Calculate:

The average score

The highest score

The lowest score


```python
scores = [60,70,100,90,80]
#avg
avg = sum(scores)/len(scores)
print(avg)
#highest
print(max(scores))
#lowest
print(min(scores))
```

Create a set of colors.

Ask the user to enter a color.

Print "Found" if the color is in the set.

Otherwise print "Not Found".

```python
colors = {"red", "blue", "green", "yellow", "orange"}
color = input("Enter a color: ")
if color in colors:
    print("Found")
else:
    print("Not Found")
```
