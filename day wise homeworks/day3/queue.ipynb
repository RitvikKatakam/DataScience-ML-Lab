2. A cinema has one ticket counter.
People arrive, stand in a queue, buy tickets, and leave.
🔹 Rules
People join the queue → enqueue(person)
Counter serves first person → dequeue()
Show next person → peek()
If queue empty → show "No customers"

```python
class CineQueue:
    def __init__(self):
        self.cine_line = []
  
    # Person joins queue
    def enqueue(self, person):
        self.cine_line.append(person)
        print(f"{person} joins the queue")
  
    # Person served by counter
    def dequeue(self):
        if self.cine_line:
            dequeued_person = self.cine_line.pop(0)
            print(f"{dequeued_person} bought ticket and left")
        else:
            print("No customers")
    
    # Show next person
    def peek(self):
        if self.cine_line:
            print(f"{self.cine_line[0]} is at the front of the queue")
        else:
            print("No customers")
```

```python
cinema = CineQueue()

cinema.enqueue("Ravi")
cinema.enqueue("Anjali")

cinema.peek()
cinema.dequeue()
cinema.peek()
cinema.dequeue()
cinema.dequeue()
```
