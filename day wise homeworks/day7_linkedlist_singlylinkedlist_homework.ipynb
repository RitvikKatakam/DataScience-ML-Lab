#Create Node Class
* task_id
* task_name
* priority
* next pointer

Perform below actions
1. Add node at begining
2. Add Node at end
3. Add in ascneding order of the priorty

```python
class Task_LinkedList:
  def __init__(self, tid, tname, priority):
    self.next = None
    self.tid = tid
    self.tname = tname
    self.priority = priority


class task_Linkedlist:
  def __init__(self):
    self.head = None

  #1
  def insert(self, tid, tname, priority):
    new_node = Task_LinkedList(tid, tname, priority)
    new_node.next = self.head
    self.head = new_node

  #2
  def insert_end(self, tid, tname, priority):
    new_node = Task_LinkedList(tid, tname, priority)

    if self.head is None:
      self.head = new_node
      return

    temp = self.head
    while temp.next:
      temp = temp.next

    temp.next = new_node

  #3
  #ascending
  def insert_asc(self, tid, tname, priority):
    new_node = Task_LinkedList(tid, tname, priority)

    if self.head is None or self.head.priority >= priority:
      new_node.next = self.head
      self.head = new_node
      return

    temp = self.head
    while temp.next and temp.next.priority < priority:
      temp = temp.next

    new_node.next = temp.next
    temp.next = new_node

  def display(self):
    temp = self.head
    while temp:
      print(f"[{temp.tid}, {temp.tname}, {temp.priority}] -> ", end="")
      temp = temp.next
    print("None")
```

```python
obj = task_Linkedlist()

obj.insert_asc(4, "Task4", 4)
obj.insert_asc(2, "Task2", 2)
obj.insert_asc(6, "Task6", 6)

obj.display()
```
