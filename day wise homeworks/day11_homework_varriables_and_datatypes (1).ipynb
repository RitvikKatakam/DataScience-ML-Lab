1. Create an array of numeric 20 values. Find the Largest Number in the array

```python
arr1 = [3,6,9,12,15,18,21,24,27,30]
largest = arr1[0]
for num in arr1:
  if num>largest:
    largest = num

print("largest number in array:",largest)
```
