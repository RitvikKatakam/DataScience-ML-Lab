Create a hashtable to store a list of employee with attributes FirstName, LastName, Salary, Address, Age with size 20

1. Write a insert function to a new employee with the above 5 attributes
2. Write a function to get a employee detail matching a specific Age
3. Write a function to delete an employee from the list with a specific Name
4. Write a function to Display all the employees with salary greater than 10000

```python
class HashTable:
    def __init__(self, size=20):
        self.size = size
        self.table = [[] for _ in range(self.size)]

    def _hash(self, key):
        return hash(key) % self.size

    #1️
    #insert
    def insert(self, fname, lname, salary, address, age):
        index = self._hash(fname)

        for i in range(len(self.table[index])):
            if self.table[index][i][0] == fname:
                print("Employee already exists. Updating...")
                self.table[index][i] = (
                    fname,
                    {
                        "lastname": lname,
                        "salary": salary,
                        "address": address,
                        "age": age
                    }
                )
                return

        self.table[index].append(
            (
                fname,
                {
                    "lastname": lname,
                    "salary": salary,
                    "address": address,
                    "age": age
                }
            )
        )

    #2
    #display
    def display(self):
        print("All Employees:")
        for index, bucket in enumerate(self.table):
            if bucket:
                print(f"\nIndex: {index}")
                for employee in bucket:
                    print(f"Name: {employee[0]}")
                    print(f"Lastname: {employee[1]['lastname']}")
                    print(f"Salary: {employee[1]['salary']}")
                    print(f"Address: {employee[1]['address']}")
                    print(f"Age: {employee[1]['age']}")

    #3
    #delete
    def delete(self, fname):
        index = self._hash(fname)

        for i in range(len(self.table[index])):
            if self.table[index][i][0] == fname:
                del self.table[index][i]
                print("Deleted employee.")
                return

        print("Employee not found.")

    #4
    #display with salary < 10000
    def display_salary_less_than_10000(self):
        print("Employees with salary less than 10000:")
        found = False

        for bucket in self.table:
            for employee in bucket:
                if employee[1]["salary"] < 10000:
                    print(f"Name: {employee[0]}")
                    print(f"Salary: {employee[1]['salary']}")
                    print("\t----\t")
                    found = True

        if not found:
            print("No employees found.")
```

```python
ht = HashTable()
```

```python
ht.insert("Rahul", "Kumar", 15000, "Hyderabad", 22)
ht.insert("Cam", "Stone", 8000, "Chennai", 25)
ht.insert("Anu", "Rao", 20000, "Mumbai", 23)
```
