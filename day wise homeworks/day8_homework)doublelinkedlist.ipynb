Create A doublist list of 5 object with fruit names
* Allow to inert it in any random order
* Write a function to sort the fruit names based on alphabatical order
* Write a function to display forward and backward nodes


```python
class Node:
  def __init__(self, data):
    self.data = data
    self.next = None
    self.prev = None


class Doublylinkedlist:
  def __init__(self):
    self.head = None

  #1
  def insert_end(self, data):
    new_node = Node(data)
    if self.head is None:
      self.head = new_node
      return

    temp = self.head
    while temp.next:
      temp = temp.next

    temp.next = new_node
    new_node.prev = temp

  #2
  def sort_alphabettically(self):
    if self.head is None:
      return

    swapped = True
    while swapped:
      swapped = False
      temp = self.head

      while temp.next:
        if temp.data > temp.next.data:
          temp.data, temp.next.data = temp.next.data, temp.data
          swapped = True
        temp = temp.next

  #3
  def display_forward(self):
    temp = self.head
    if temp is None:
      print("List is empty")
      return

    while temp:
      print(temp.data, end=" <=> ")
      temp = temp.next
    print("None")

  def display_backward(self):
    temp = self.head
    if temp is None:
      print("List is empty")
      return

    # move to last node
    while temp.next:
      temp = temp.next

    while temp:
      print(temp.data, end=" <=> ")
      temp = temp.prev
    print("None")
```

```python
dll = Doublylinkedlist()
dll.insert_end("Apple")
dll.insert_end("Mango")
dll.insert_end("Banana")
dll.insert_end("Grapes")
dll.insert_end("Orange")

print("Before Sorting:")
dll.display_forward()

dll.sort_alphabettically()

print("\nAfter Sorting (Forward):")
dll.display_forward()

print("\nAfter Sorting (Backward):")
dll.display_backward()
```
