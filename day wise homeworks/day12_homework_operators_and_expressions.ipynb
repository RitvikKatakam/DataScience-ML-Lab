##1. Write a program to calculate the square and cube of a number.

```python
print("1. square")
print("2. cube")

choice = int(input("Enter your choice: "))
num = int(input("Enter a number: "))

if choice == 1:
    print("Square is:", num**2)

elif choice == 2:
    print("Cube is:", num**3)

else:
    print("Invalid choice")
```

## 2. Write a Python program to calculate Simple Interest.

```python
principle = int(input("enter the value for principle:"))
time = int(input("enter the value for time:"))
rate = int(input("enter the value for rate:"))

si = (principle*rate*time)/100

print("simple interest:",si)
```

# 3. Employee Bonus Calculator

Write a Python program that calculates the employee bonus based on the employee's salary.

## Rules to Follow

1. If Salary > 50000 → Bonus = 10% of salary

2. If Salary is between 30000 and 50000 → Bonus = 5% of salary

3. If Salary < 30000 → Bonus = 2% of salary



```python
salary = int(input("enter the value for employee salary:"))

if salary > 50000:
  bonus = (salary*10)//100
  print("bonus:",bonus)
  print("total salary:",salary+bonus)

elif salary >= 30000 and salary <= 50000:
  bonus = (salary*5)//100
  print("bonus:",bonus)
  print("total salary:",salary+bonus)

elif salary < 30000:
  bonus = (salary*2)//100
  print("bonus:",bonus)
  print("total salary:",salary+bonus)

else:
  print("invalid input")
```

# 4. Electricity Bill Calculator

Write a Python program to calculate the electricity bill based on the number of units consumed.

## Rules to Follow

1. If Units ≤ 100 → Charge ₹5 per unit

2. If Units between 101 and 200 → Charge ₹7 per unit

3. If Units > 200 → Charge ₹10 per unit

```python
units = int(input("enter the value for units:"))

if units <= 100:
  charge = units*5
  print("charge:",charge)
  print("total bill:",charge)
elif units >= 101 and units <= 200:
  charge = units*7
  print("charge:",charge)
  print("total bill:",charge)
elif units > 200:
  charge = units*10
  print("charge:",charge)
  print("total bill",charge)
```
