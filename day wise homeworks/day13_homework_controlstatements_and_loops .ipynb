## Write a program to check whether a year is a leap year or not

```python
year = int(input("enter a yesr:"))
if (year%4==0 and year%100!=0) or (year%400==0):
  print(f"{year} is a leap year")
else:
  print(f"{year} is not a leap year")
```

##Write a program that stops printing numbers when the number 7 appears using break

```python
n = int(input("enter a number: "))
for i in range(n):
  if i == 7:
    break
  print(i)
```

##Find the sum of all odd numbers between 1 and 100

```python
sum = 0

for i in range(1,101):
  if i%2 != 0:
    sum+=i

print("sum of odd numbers is",sum)
```

## Number Guessing Game

# This program:

* The computer chooses a number between 1 and 10

* The user guesses the number

* The program keeps asking until the correct guess

* Uses while loop and break

```python
import random

num = random.randint(1,10)
guess = 0

while guess != num:
  guess = int(input("Guess a number between 1 and 10: "))
  if guess == num:
    print("correct guess")
    break
  else:
    print("try again")
```
