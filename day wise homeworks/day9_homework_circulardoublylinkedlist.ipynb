Using Doubly circular LinkedList, creat nodes in such a way that each Node consist below attributes


* song_name
* artist
* duration
* prev
* next

Write a function to display next music information

Write a function to display previous music information

```python
class Node:
  def __init__(self, songname, artist, duration):
    self.songname = songname
    self.artist = artist
    self.duration = duration
    self.prev = None
    self.next = None


class Music_player_list:
  def __init__(self):
    self.head = None
    self.current = None

  #insert a song
  def add_song(self, songname, artist, duration):
    new_node = Node(songname, artist, duration)

    if self.head is None:
      self.head = new_node
      self.current = new_node
      new_node.next = new_node
      new_node.prev = new_node
      return

    last = self.head.prev

    new_node.prev = last
    last.next = new_node

    new_node.next = self.head
    self.head.prev = new_node

  #display current song
  def display_current(self):
    if self.current:
      print("Now playing")
      print(self.current.songname, "-", self.current.artist, "-", self.current.duration)

  #display next
  def display_next(self):
    if self.current:
      self.current = self.current.next
      print("next song")
      print(self.current.songname, "-", self.current.artist, "-", self.current.duration)

  #display previous
  def display_previous(self):
    if self.current:
      self.current = self.current.prev
      print("previous song")
      print(self.current.songname, "-", self.current.artist, "-", self.current.duration)
```

```python
playlist = Music_player_list()

playlist.add_song("beykayali", "Sachet–Parampara","6:11")
playlist.add_song("Toofan", "Ravi-Basoori","3:35")
playlist.add_song("song 3","artist 3","duration 3")


playlist.display_current()
playlist.display_next()
playlist.display_next()
playlist.display_previous()
playlist.display_previous()
```
